```python
from collections import deque
queue = deque()

queue.append(1)
queue.append(2)
queue.append(3)
queue.append(4)

print(queue)  # [2,3,4]
```

```python
queue.pop()  # 移除队列的第一个元素
print(queue)  # [3,4]
```
